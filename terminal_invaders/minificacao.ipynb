```python
# 1. Lê o arquivo original gigante
with open(r'C:\Users\Wellington.Barros\Documents\VSCode\personal_proj\terminal_invaders\texte.sql') as f:
    raw_query = f.read()

# 2. Engenharia da String: Substitui múltiplas quebras de linha e espaços por um único espaço
# Isso não altera a lógica do SQL, apenas o transforma em uma linha única e contínua
minified_query = re.sub(r"\s+", ' ', raw_query).strip()
minified_query = re.sub(r', ', '', minified_query).strip()
minified_query = re.sub(r' \( | \(|\( ' , r'\(', minified_query).strip()
minified_query = re.sub(r' \) | \)|\) ' , r'\)', minified_query).strip()
minified_query = re.sub(r'> ' , r'>', minified_query).strip()
minified_query = re.sub(r'< ' , r'<', minified_query).strip()
minified_query = re.sub(r' \"|\" ' , r'\"', minified_query).strip()
minified_query = re.sub(r' \'|\' ' , r'\'', minified_query).strip()
minified_query = re.sub(r'= | =' , r'=', minified_query).strip()
minified_query = re.sub(r'\\' , '', minified_query).strip()

# minified_query = minified_passo_1
# 3. Validação do tamanho
print(f"Tamanho original: {len(raw_query)} caracteres")
# print(f"Tamanho comprimido: {len(minified_passo_1)} caracteres")
# print(f"Tamanho comprimido: {len(minified_passo_2)} caracteres")
print(f"Tamanho comprimido: {len(minified_query)} caracteres")
```
